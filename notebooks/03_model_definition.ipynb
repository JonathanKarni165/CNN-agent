# 03 - Model Definition
Every model implements the same `BaseModel` interface (`src/models.py`) — same Strategy pattern as the features — so notebook 04 can train/compare them identically. This notebook is just where you configure which models and which loss function to try.

```python
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().parent
sys.path.insert(0, str(PROJECT_ROOT))

from src.models import RidgeModel, XGBoostModel, XGB_OBJECTIVES

print('Available XGBoost loss functions:', list(XGB_OBJECTIVES.keys()))
```

## Tier 1 - Ridge (fixed squared-error loss + L2)

```python
ridge_model = RidgeModel(alpha=1.0)
ridge_model.name
```

## Tier 2 - XGBoost
Set `XGB_LOSS` to `'squared'`, `'mae'`, or `'huber'` (robust to fat-tailed jumps, matches the spec's default) to change the training objective.

```python
XGB_LOSS = 'huber'

xgboost_model = XGBoostModel(
    loss=XGB_LOSS,
    n_estimators=500,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.7,
    colsample_bytree=0.7,
    early_stopping_rounds=30,
)
xgboost_model.name
```

## Models to run
Notebook 04 reads this dict — add/remove entries to change what gets trained & compared.

```python
MODELS = {
    'ridge': ridge_model,
    'xgboost': xgboost_model,
}
list(MODELS.keys())
```
